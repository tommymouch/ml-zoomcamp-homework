```python
df_2cols = df_asia[['vehicle_weight', 'model_year']].head(7)
df_2cols.head()
```

```python
X = df_2cols.to_numpy()
print(X)
```

```python
TX =X.T
print(TX)
```

```python
XTX = TX.dot(X)
print(XTX)
```

```python
import numpy as np
```

```python
inv_XTX = np.linalg.inv(XTX)
print(inv_XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = inv_XTX.dot(TX).dot(y)
```
